```python
import pandas as pd
```

```python
df_2021 = pd.read_csv("world-happiness-report-2021.csv")
df_2021.columns = (
    df_2021.columns
    .str.strip()                          # Supprime les espaces au début/fin
    .str.lower()                          # Tout en minuscules
    .str.replace(' ', '_')                # Remplace les espaces par des underscores
    .str.replace(r'[^\w]', '', regex=True) # Supprime les caractères spéciaux
)

#doublon c'est tchao
df_2021 = df_2021.drop_duplicates()

# Identifier les NaN
print(df_2021.isna().sum())
# Supprimer ou imputer selon le cas
#df_2021 = df_2021.dropna(subset=['colonne_critique'])

df_2021.info()

df_2021.head()
```
